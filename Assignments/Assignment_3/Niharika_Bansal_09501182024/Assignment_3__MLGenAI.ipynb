Ques 2. Create a function to calculate the sum for first N natural numbers

```python
def Sum_no(N):
  sum_no = 0
  for i in range (1, N+1):
    sum_no = sum_no + i

  return sum_no

N = int(input("Enter no of terms N= "))
print(Sum_no(N))

```

Ques 3. Create a function to reveres a number

```python
n = int(input("Enter your number: "))

def Rev_num(num):
  rev = 0
  while num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num //= 10
  return rev

print("Reverse of the number is ", Rev_num(n))
```

Ques 5. Create a function to check palindrome number

```python
n = int(input("Enter your number: "))

def Palindrome(num):
  rev = 0
  while num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num //= 10
  return rev == num

if Palindrome(n):
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

Ques 6. Create a function to generate Fibonacci Series

```python
def Fib(n):
  a = 0
  b = 1
  for i in range (n):
    print(a)
    c = a + b
    a = b
    b = c

n = int(input("Enter no of terms: "))
Fib(n)
```

Ques 11. Create a Student Class with Name and Marks

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Student Name:", self.name)
        print("Marks:", self.marks)

name = input("Enter student name: ")
marks = float(input("Enter marks: "))
s1 = Student(name, marks)
s1.display()
```
