Ques 1. Find area of Rectangle

```python
Length = float(input("Enter Length "))
Breadth = float(input("Enter Breadth "))
area = Length * Breadth
print("Area of the Rectangle =", area)
```

Ques 2. Find Simple Interest

```python
Principal = float(input("Enter Principal Amount "))
Rate = float(input("Enter Rate of Interest "))
Time = float(input("Enter Time duration in Years "))

Simple_Interest = ( Principal * Rate * Time ) / 100

print("Simple Interest for your amount is ", Simple_Interest)
```

Ques 3. Convert Temp from Celsius to Fahrenhiet

```python
Temp_c = float(input("Enter Temp in Celsius "))
Temp_f = 9/5 * Temp_c + 32
print("Temperature in Fahrenhiet is ", Temp_f)
```

Ques 4. Calculate Average of 3 numbers

```python
No1 = float(input("Enter no 1. "))
No2 = float(input("Enter no 2. "))
No3 = float(input("Enter no 3. "))

avg = (No1 + No2 + No3 ) / 3

print("Average is ", avg)
```

Ques 5. Find square and cube of no

```python
num = float(input("Enter number "))
sq = num ** 2
cube = num ** 3
print("Square is", sq)
print("Cube is", cube)
```

Ques 6. Swap 2 variables without 3rd

```python
a = int(input("enter value of a "))
b = int(input("enter value of b "))
a, b = b, a
print("After Swap; a = ", a)
print("b = ", b)
```

Ques 7. Student Report Program

```python
# Input Student Details
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

# Input Marks
english = float(input("Enter English Marks: "))
maths = float(input("Enter Maths Marks: "))
science = float(input("Enter Science Marks: "))
hindi = float(input("Enter Hindi Marks: "))
sst = float(input("Enter SST Marks: "))

# Calculate Total
total = english + maths + science + hindi + sst

# Calculate Percentage
percentage = (total / 500) * 100

# Display Report
print("\n----- STUDENT REPORT -----")
print("Student Name :", name)
print("Roll Number  :", roll_no)

print("\nMarks")
print("English :", english)
print("Maths   :", maths)
print("Science :", science)
print("Hindi :", hindi)
print("SST :", sst)

print("\nTotal Marks :", total)
print("Percentage  :", percentage, "%")
```
